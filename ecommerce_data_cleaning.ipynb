```python
#12.Aşağıdaki koşullara uyan mantıksal olarak hatalı kayıtları bulun: birim_fiyat <= 0, toplam_tutar <= 0, teslimat_gunu < 0, musteri_puani > 5. Bu kayıtların sayısını yazdırın ve veri setinden çıkarın.
```

```python
df = df_original.copy()
hatali_kayitlar=df[
    (df['birim_fiyat']<=0)|
    (df['toplam_tutar']<=0)|
    (df['teslimat_gunu']<0)|
    (df['musteri_puani']>5)
]

pd.set_option('display.max_columns', None)
print(f"Hatalı kayıt sayısı: {len(hatali_kayitlar)}")
print(f"Temizlemeden önce: {df.shape}")

df = df[
    (df['birim_fiyat'] > 0) &
    (df['toplam_tutar'] > 0) &
    (df['teslimat_gunu'] >= 0) &
    (df['musteri_puani'] <= 5)
]

print(f"Temizledikten sonra: {df.shape}")
```

Bu bölümde mantıksal olarak hatalı kayıtları temizlerken **beklenmedik bir sonuç** ile karşılaştım

Bu kodun çıktısı beklediğimden farklıydı:

- **Hatalı kayıt sayısı:** 2
- **Temizlemeden önce:** 1428 satır
- **Temizledikten sonra:** 1342 satır

Matematiksel olarak `1428 - 2 = 1426` olması gerekirken **1342** çıktı. **84 satır fazla** silinmişti.

Araştırdığımda, filtreleme yaptığım sütunlarda **84 NaN değeri** olduğunu fark ettim:

```python
print("teslimat_gunu NaN sayısı:", df_original['teslimat_gunu'].isna().sum())
print("musteri_puani NaN sayısı:", df_original['musteri_puani'].isna().sum())
print("birim_fiyat NaN sayısı:", df_original['birim_fiyat'].isna().sum())
print("toplam_tutar NaN sayısı:", df_original['toplam_tutar'].isna().sum())
```

Pandas'ta bir koşul yazdığında (df['sütun'] > 0 gibi), Pandas her satır için True veya False üretir.

df[koşul] derken → sadece True olanları tutar.

NaN "bilinmiyor" demek. Pandas "5'ten küçük mü büyük mü bilmiyorum" der ve False döner.

Pandas'ta NaN değerler karşılaştırma işlemlerinde her zaman False döner.

- `NaN > 0` → False
- `NaN >= 0` → False
- `NaN <= 5` → False

Bu yüzden `df[koşul]` yaptığımda NaN olan tüm satırlar filtreye takıldı ve yanlışlıkla silindi.

```python
df = df_original.copy()

# Hatalı koşulu tanımla
hatali = (
    (df['birim_fiyat'] <= 0) |
    (df['toplam_tutar'] <= 0) |
    (df['teslimat_gunu'] < 0) |
    (df['musteri_puani'] > 5)
)

print(f"Hatalı kayıt sayısı: {hatali.sum()}")
print(f"Temizlemeden önce: {df.shape}")

# Hatalı olmayanları tut (NaN'lar korunur)
df = df[~hatali]

print(f"Temizledikten sonra: {df.shape}")
```


### Neden NaN'ları Silmek İstemedim?

NaN "yanlış değer" değil, **"eksik bilgi"**. Bir müşterinin puan vermemiş olması, siparişinin, adresinin, kategori bilgisinin de yanlış olduğu anlamına gelmiyor. Bu 84 kaydı tamamen silmek diğer değerli bilgileri de kaybetmek olurdu.

Doğru yaklaşım: mantıksız değerleri bu aşamada silmek, NaN değerleri **sonraki aşamada** ayrıca ele almak (fillna ile doldurma).

Bu yaklaşımda NaN değerler:
- `hatali` koşulunda **False** döner (hatalı değil sayılır)
- `~hatali` ile **True** olup korunur

Bu iki durumu karıştırmamak için "hatalı koşulu tanımla, `~` ile tersini al" yaklaşımı daha güvenlidir.

```python
#13.birim_fiyat sütunu için IQR yöntemini kullanarak Q1, Q3, IQR, alt sınır ve üst sınır değerlerini hesaplayın. Sonra aykırı değer içeren kayıt sayısını yazdırın.
```

```python
df = df_original.copy()

Q1 = df['birim_fiyat'].quantile(0.25)
Q3 = df['birim_fiyat'].quantile(0.75)
IQR = Q3 - Q1

alt_sinir = Q1 - 1.5 * IQR
ust_sinir = Q3 + 1.5 * IQR

print(f"Q1 : {Q1:.2f}")
print(f"Q3 : {Q3:.2f}")
print(f"IQR: {IQR:.2f}")
print(f"Alt sınır: {alt_sinir:.2f}")
print(f"Üst sınır: {ust_sinir:.2f}")

aykiri_degerler = df[(df['birim_fiyat'] < alt_sinir) | (df['birim_fiyat'] > ust_sinir)]
print(f"\nAykırı değer içeren kayıt sayısı: {len(aykiri_degerler)}")
```

```python
#14.toplam_tutar sütunu için histogram çizdirin ve box plot çizdirin. Ardından dağılım hakkında 2–3 cümlelik kısa bir yorum yazın.
```

```python
df = df_original.copy()

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Histogram
axes[0].hist(df['toplam_tutar'].dropna(), bins=30, color='indianred' , edgecolor='black')
axes[0].set_title("Toplam Tutar - Histogram")
axes[0].set_xlabel("Toplam Tutar (TL)")
axes[0].set_ylabel("Frekans")
axes[0].grid(True)

# Boxplot
axes[1].boxplot(df['toplam_tutar'].dropna(), vert=False)
axes[1].set_title("Toplam Tutar - Boxplot")
axes[1].set_xlabel("Toplam Tutar (TL)")
axes[1].grid(True)

plt.tight_layout()
plt.show()

# İstatistikler
print(f"Mean: {df['toplam_tutar'].mean():.2f}")
print(f"Median: {df['toplam_tutar'].median():.2f}")
print(f"Std: {df['toplam_tutar'].std():.2f}")
```

```python
#15.Aşağıdaki analizleri yapın: kategori sütunu için value_counts() çıktısını yazdırın,
#musteri_tipi sütunu için yüzde dağılımı hesaplayın,
#odeme_turu sütunu için count plot çizdirin,
#birim_fiyat, toplam_tutar ve musteri_puani sütunlarının temel istatistiklerini çıkarın.
#Son olarak şu iki soruya cevap verin: Hangi kategori en yüksek ortalama toplam_tutar değerine sahip?
#Hangi müşteri tipi en yüksek ortalama musteri_puani değerine sahip?
```

```python
#Aşağıdaki analizleri yapın: kategori sütunu için value_counts() çıktısını yazdırın
df = df_original.copy()
df['kategori'] = (
    df['kategori']
    .str.strip()
    .str.lower()
    .str.replace('&', ' ve ')
    .str.replace('ı', 'i')
    .str.replace(r'\s+', ' ', regex=True)
    .str.title()
)
print("-------Kategori Dağılımı:--------\n")
print(df['kategori'].value_counts())
```

```python
#musteri_tipi sütunu için yüzde dağılımı hesaplayın
df = df_original.copy()
df['musteri_tipi'] = df['musteri_tipi'].str.strip().str.lower().str.title()
print("\n----------Müşteri Tipi Yüzde Dağılımı:-----------")
print(df['musteri_tipi'].value_counts(normalize=True) * 100)
```

```python
#odeme_turu sütunu için count plot çizdirin
df = df_original.copy()
print("------------Ödeme Türü Dağılımı-----------")
def temizle(sutun):
    return (
        sutun
        .str.strip()
        .str.lower()
        .str.replace('ı', 'i')
        .str.replace('İ', 'i')
        .str.replace('ğ', 'g')
        .str.replace('ü', 'u')
        .str.replace('ş', 's')
        .str.replace('ö', 'o')
        .str.replace('ç', 'c')
        .str.replace('&', ' ve ')
        .str.replace('/', ' ')
        .str.replace('-', ' ')
        .str.replace('_', ' ')
        .str.replace('.', ' ')
        .str.replace(',', ' ')
        .str.replace(r'\s+', ' ', regex=True)
        .str.strip()
        .str.title()
    )

df = df_original.copy()
df['odeme_turu'] = temizle(df['odeme_turu'])
df['kategori'] = temizle(df['kategori'])
df['musteri_tipi'] = temizle(df['musteri_tipi'])

# Kısaltmalar için manuel eşleştirme
esitleme_odeme = {
    'Kart': 'Kredi Karti',
    'Eft': 'Havale Eft',
    'Havale': 'Havale Eft',
}
df['odeme_turu'] = df['odeme_turu'].replace(esitleme_odeme)

plt.figure(figsize=(10, 5))
sns.countplot(data=df, x='odeme_turu', color='#E76F51' , edgecolor='black')
plt.title("Ödeme Türü Dağılımı")
plt.xlabel("Ödeme Türü")
plt.ylabel("Adet")
plt.xticks(rotation=45)
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```
